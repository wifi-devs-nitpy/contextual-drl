```python
from flax import linen as nn 

import jax
import jax.numpy as jnp
import flax.linen as nn

# 1. Define the network as a class inheriting from nn.Module
class Network(nn.Module):
    @nn.compact
    def __call__(self, x):
        x = nn.Dense(64)(x)
        x = nn.relu(x)
        x = nn.Dense(64)(x)
        x = nn.relu(x)
        return nn.Dense(5)(x)

model = Network()

x = jnp.ones((1, 32))  # Example shape
key = jax.random.PRNGKey(0)
variables = model.init(key, x)
# 4. Run the model using model.apply()
output = model.apply(variables, x)
```
