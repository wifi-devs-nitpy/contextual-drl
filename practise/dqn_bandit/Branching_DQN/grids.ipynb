```python
import jax.numpy as jnp 

a = jnp.arange(10)
b = jnp.arange(3)

grids = jnp.meshgrid(a, b, indexing="ij")   
```

```python
sta_link = {
    (int(sta), int(link)): int(sta + link)

    for sta, link in jnp.stack([grid.ravel() for grid in grids], axis=-1)
}
```

```python
print(sta_link)
```

```python
sta_link[jnp.int32(3), 2]
```

```python
for s_l in jnp.stack([grid.ravel() for grid in grids], axis=-1):
    print(tuple(s_l.tolist()))
```

```python
import numpy as np 
from chex import Array 
```

```python
access_points = [4, 8, 10, 12, 16]
selected_ap_group = (10, 4, 12)
```

```python
def _encode_ap_group(sharing_ap, selected_ap_group) -> Array: 
    ap_group = np.asarray(list(selected_ap_group) + [sharing_ap])
    print(ap_group)
    return (
        np.isin(
            np.asarray(access_points), 
            ap_group
        ).astype(np.int32)
    )
```

```python
_encode_ap_group(sharing_ap=10, selected_ap_group=selected_ap_group)
```
